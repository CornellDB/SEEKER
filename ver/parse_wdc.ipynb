```python
# help me parse WDC_data

import csv
import json
import re
import tqdm

dir = "WDC_data"

with open('sample', 'r') as f:
    for line in f:
        data_dict = json.loads(line)
        transposed = list(zip(*data_dict["relation"]))

        filename = re.sub(r'[\\/*?:"<>|]', "", data_dict["pageTitle"])[:8] + ".csv"

        with open(dir + "/" + filename, 'w', newline='') as csvfile:
            writer = csv.writer(csvfile)
            writer.writerows(transposed)


```
